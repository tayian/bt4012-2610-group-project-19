### Test Gemini API

```python
# 1. Initialize Gemini Client
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()
api_key = os.getenv("GEMINI_API_KEY")
print(f"API Key found: {bool(api_key)}")

client = genai.Client(api_key=api_key)
print("Gemini Client initialized!")
```

```python
# 2. Test Content Generation
response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents="Hello Gemini! Explain utility fraud detection in two sentences."
)

print("Gemini Response:")
print(response.text)
```

```python
# 3. Test Structured Output with Pydantic (For Role D Inspector Briefs)
from pydantic import BaseModel
from typing import List

class InspectionBrief(BaseModel):
    client_id: str
    risk_tier: str
    suspected_patterns: List[str]
    onsite_action: str

structured = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Client train_Client_001 shows an 85% drop in consumption and negative meter index progression.",
    config={
        "response_mime_type": "application/json",
        "response_schema": InspectionBrief,
        "temperature": 0.0
    }
)

print("Structured JSON Output:")
print(structured.text)
```
